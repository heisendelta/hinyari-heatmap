# Data Preprocessing

For data from 2024, group by month, then scale by the number of cases observed in a month (region df)<br>
For data from 2023 and earlier, group by year, then scale by the number of cases observed in a year (region df)

### 2024

```python
df_2024 = pd.DataFrame()

for i in range(1, 7):
    region_df = pd.read_csv(f'data/region/region_2024_{i}.csv')
    region_df['month'] = i
    df_2024 = pd.concat([df_2024, region_df], axis=0)
```

```python
na_count = df_2024.groupby(['month', '市区町丁'])['非侵入窃盗計'].sum().unstack().isna().sum().sort_values(ascending=False)
df_2024_timeseries = df_2024.groupby(['month', '市区町丁'])['非侵入窃盗計'].sum().unstack().T.loc[na_count.index[na_count == 0]]
```

## 2023 and earlier

```python
df_2023earlier = pd.DataFrame()

for i in range(2019, 2023):
    region_df = pd.read_csv(f'data/region/region_{i}.csv')
    region_df['year'] = i
    df_2023earlier = pd.concat([df_2023earlier, region_df], axis=0)
```

```python
na_count_2023earlier = df_2023earlier.groupby(['year', '市区町丁'])['非侵入窃盗計'].sum().unstack().isna().sum().sort_values(ascending=False)
df_2023earlier_timeseries = df_2023earlier.groupby(['year', '市区町丁'])['非侵入窃盗計'].sum().unstack().T.loc[na_count_2023earlier.index[na_count_2023earlier == 0]]
```
