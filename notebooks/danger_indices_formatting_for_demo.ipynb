```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
```

```python
influenced_series_df = pd.read_csv('final_data/influenced_series.csv')
with open('final_data/predict_dicts.pkl', 'rb') as f:
    predict_dicts = pickle.load(f)
```

```python
influenced_series_df = influenced_series_df.set_index('市区町丁', drop=True)
influenced_series_df.head()
```

```python
danger_indices_list = []

for index, row in influenced_series_df.iterrows():
    neighbors = list(predict_dicts[row.name].keys())
    mean_values = influenced_series_df.loc[neighbors].mean()
    danger_indices_list.append(mean_values)

danger_indices = pd.DataFrame(danger_indices_list, index=influenced_series_df.index, columns=influenced_series_df.columns)
```
